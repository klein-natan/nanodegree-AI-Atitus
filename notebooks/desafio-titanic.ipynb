## Os dados

| Coluna | O que é |
|---|---|
| `id_passageiro` | o número do passageiro (não é característica: não use no modelo) |
| `classe` | 1, 2 ou 3: a classe da passagem |
| `sexo` | `feminino` ou `masculino` |
| `idade` | em anos |
| `irmaos_conjuge` | quantos irmãos ou cônjuges viajavam junto |
| `pais_filhos` | quantos pais ou filhos viajavam junto |
| `tarifa` | quanto pagou pela passagem |
| `porto_embarque` | `Southampton`, `Cherbourg` ou `Queenstown` |
| `sobreviveu` | **a resposta**: 1 se sobreviveu (só no treino) |

## O arquivo de envio

O exemplo abaixo usa a régua (ninguém sobrevive), só para você testar o
envio no placar. Troque as duas primeiras linhas pelas previsões do seu
modelo. Depois rode a célula: o Colab baixa o arquivo `envio.csv`, e é

```
# Troque estas duas linhas pelas previsões do seu modelo
probabilidades = np.zeros(len(teste))
previsoes = (probabilidades >= 0.5).astype(int)

envio = pd.DataFrame({
    "id_passageiro": teste["id_passageiro"],
    "sobreviveu": previsoes,
    "probabilidade": probabilidades,
})
envio.to_csv("envio.csv", index=False)
print(envio.head())
print("Linhas:", len(envio), "(precisam ser 300)")
```
